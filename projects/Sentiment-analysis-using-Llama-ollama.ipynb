```python
That sounds like a great way to measure customer sentiment progression throughout a support chat. Here’s how you can approach it:
    
    🔹 Approach
Extract first and last customer utterances from the chat.
Classify sentiment for both the first and last utterance.
Compare sentiment scores to determine if the sentiment improved, remained the same, or worsened.
```

```python
import ollama
import json

# Example support chat (Only customer messages)
chat_history = [
    "I'm really frustrated! My internet is down again.",
    "I've tried restarting the router but it didn't help.",
    "Okay, I will follow the troubleshooting steps.",
    "Thank you, it's working now. Appreciate your help!"
]

# Select first and last utterance from chat
first_utterance = chat_history[0]
last_utterance = chat_history[-1]

# Constructing a structured few-shot prompt for sentiment analysis
prompt = f"""<|system|>
You are an AI that analyzes customer sentiment progression in a support chat.
For both the first and last utterance, return:
- **Sentiment**: Positive, Neutral, or Negative
- **Confidence score** (between 0 and 1)
- **Emotion** (e.g., Happy, Angry, Sad, Frustrated, Relieved, Grateful)

Respond in **JSON format** as:
{{
    "sentiment_progression": {{
        "first_utterance": {{"sentence": "...", "sentiment": "...", "emotion": "...", "score": ...}},
        "last_utterance": {{"sentence": "...", "sentiment": "...", "emotion": "...", "score": ...}},
        "overall_change": "Improved / Worsened / Same"
    }}
}}

### **Examples:**

First Utterance: "I'm really annoyed with this long wait time!"
Sentiment: Negative
Emotion: Frustrated
Score: 0.15

Last Utterance: "Thanks a lot for your help!"
Sentiment: Positive
Emotion: Grateful
Score: 0.90

Overall Change: Improved

Now, analyze the following chat:

First Utterance: "{first_utterance}"
Last Utterance: "{last_utterance}"
"""

# Call Ollama API
response = ollama.chat(model="llama3", messages=[{"role": "user", "content": prompt}])

# Parse and print response
output = json.loads(response["message"]["content"])
print(json.dumps(output, indent=2))
```

```python
{
  "sentiment_progression": {
    "first_utterance": {
      "sentence": "I'm really frustrated! My internet is down again.",
      "sentiment": "Negative",
      "emotion": "Frustrated",
      "score": 0.20
    },
    "last_utterance": {
      "sentence": "Thank you, it's working now. Appreciate your help!",
      "sentiment": "Positive",
      "emotion": "Grateful",
      "score": 0.85
    },
    "overall_change": "Improved"
  }
}
```

✅ LLaMA

- Can be fine-tuned for sentiment analysis using LoRA/PEFT
- Efficient and supports large-scale sentiment understanding

- Not explicitly trained for sentiment analysis but can infer sentiment through zero-shot or few-shot prompting.
- More flexible—can handle nuanced sentiment, multi-class sentiment, and contextual understanding.
- Slower and resource-intensive compared to DistilBERT.
- Needs fine-tuning or prompt engineering to match DistilBERT's accuracy for structured sentiment tasks.

- Using LLaMA (Large Language Model Meta AI) for sentiment analysis requires a different approach compared to transformer-based classifiers like BERT or DistilBERT. Since LLaMA is a generative model, it isn't fine-tuned for classification by default, but you can still use it effectively for sentiment analysis.
